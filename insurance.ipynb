```python
import pandas as pd
```

```python
data = pd.read_csv("insurance.csv")
```

```python
data.head()
```

```python
data.info()
```

```python
data.isnull().sum()
```

```python
data.duplicated().sum()
```

```python
data.drop_duplicates(inplace=True)
```

```python
data.duplicated().sum()
```

```python
data["sex"] = data["sex"].replace({
    "male": 1,
    "female": 0
})
```

```python
data["smoker"] = data["smoker"].replace({
    "yes": 1,
    "no": 0
})
```

```python
data = pd.get_dummies(data, columns=["region"], dtype=int)
```

```python
data.head()
```

```python
X = data.drop("charges", axis=1)
y = data["charges"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=32
)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
Insurance_predictor = LinearRegression()
```

```python
Insurance_predictor.fit(X_train, y_train)
```

```python
y_predict = Insurance_predictor.predict(X_test)
```

```python
from sklearn.metrics import mean_absolute_error, r2_score
```

```python
mae = mean_absolute_error(y_test, y_predict)

print("Mean Absolute Error:", mae)
```

```python
r2 = r2_score(y_test, y_predict)

print("R² Score:", r2)
```

```python
def predict_charges(age, sex, bmi, children, smoker, region):

    sex = 1 if sex == "male" else 0
    smoker = 1 if smoker == "yes" else 0

    new_data = pd.DataFrame({
        "age": [age],
        "sex": [sex],
        "bmi": [bmi],
        "children": [children],
        "smoker": [smoker],
        "region_northeast": [1 if region == "northeast" else 0],
        "region_northwest": [1 if region == "northwest" else 0],
        "region_southeast": [1 if region == "southeast" else 0],
        "region_southwest": [1 if region == "southwest" else 0]
    })

    prediction = Insurance_predictor.predict(new_data)

    return prediction[0]
```

```python
result = predict_charges(
    30,
    "male",
    28.5,
    1,
    "no",
    "southwest"
)

print("Predicted Insurance Charges:", result)
```
